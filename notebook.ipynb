## **3. Data Preparation / Preprocessing**

### 3.1 Menghapus Kolom yang Tidak Relevan

### 3.2 Pengecekan Outlier

```python
Q1 = df_bersih['MonthlyIncome'].quantile(0.25)
Q3 = df_bersih['MonthlyIncome'].quantile(0.75)
IQR = Q3 - Q1

batas_bawah = Q1 - 1.5 * IQR
batas_atas = Q3 + 1.5 * IQR

outliers_income = df_bersih[(df_bersih['MonthlyIncome'] < batas_bawah) | (df_bersih['MonthlyIncome'] > batas_atas)]
print(f"Jumlah baris yang terdeteksi sebagai outlier pada MonthlyIncome: {len(outliers_income)}")
```

### 3.3 Memisahkan Data untuk Analisis dan Prediksi

## **4. Modeling**

### 4.1 Encoding

```python
# Memisahkan feature (X) dengan target (y)
X = df_train.drop(columns=['Attrition'])
y = df_train['Attrition'].astype(int)     # Memastikan target berupa int

# One-Hot Endcoding
X_encoded = pd.get_dummies(X, drop_first=True)

print("Tahap 4.1 Selesai!")
print("Jumlah kolom fitur sekarang setelah di-encode:", X_encoded.shape[1])
```

### 4.2 Train-Test Split

```python
from sklearn.model_selection import train_test_split

# 80 train : 20 test
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, 
    y, 
    test_size=0.2, 
    random_state=42, 
    stratify=y # Memastikan proporsi karyawan resign dan tidak resign tetap seimbang
)

print("Tahap 4.2 Selesai!")
print("Jumlah baris data latih (X_train):", len(X_train))
print("Jumlah baris data uji (X_test):", len(X_test))
```

```python
# Mengambil 1 baris pertama dari data X_test dalam bentuk dictionary untuk uji input di prediction.py
print(X_test.head(1).to_dict(orient='records')[0])
```

### 4.3 Melatih Model Machine Learning

```python
from sklearn.ensemble import RandomForestClassifier

# Membuat otak ML
# class_weight="balanced" digunakan karena jumlah karyawan resign lebih sedikit
rf_model = RandomForestClassifier(n_estimators=100, random_state=42, class_weight="balanced")

# Training
rf_model.fit(X_train, y_train)

print("Tahap 4.3 Selesai! Model berhasil dilatih dan siap diuji.")
```

## **5.Evaluation**

### 5.1 Classification Report)

```python
from sklearn.metrics import classification_report, confusion_matrix

# Meminta model menebak data uji
y_pred = rf_model.predict(X_test)

# Menampilkan nilai rapor (Akurasi, Presisi, Recall)
print("Laporan Evaluasi Model (Classification Report):")
print(classification_report(y_test, y_pred))
```

### 5.2 Memvisualisasikan Hasil Tebakan (Confusion Matrix)

```python
plt.figure(figsize=(6, 4))
cm = confusion_matrix(y_test, y_pred)

# Grafik heatmap
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=['Bertahan (0)', 'Resign (1)'],
            yticklabels=['Bertahan (0)', 'Resign (1)'])

plt.title('Confusion Matrix - Hasil Tebakan Model', fontsize=14, pad=15)
plt.xlabel('Prediksi Mesin')
plt.ylabel('Kenyataan (Aktual)')
```

**Evaluasi Model Machine Learning**<br>
Algoritma Random Forest memberikan tingkat akurasi sebesar 84%. Berdasarkan heatmap *Confustion Matrix* dapat beberapa insight yang didapatkan bahwa :
- Model menebak 175 karyawan akan bertahan (tebakan benar)
- Model menebak 4 karyawan akan *resign* (tebakan benar)
- Model menebak 32 karyawan akan bertahan (tebakan salah/false negative)
- Model menebak 1 karyawan akan *resign* (tebakan salah/false positive)

Pada dasarnya Random Forest sangat akurat dalam mengenali karyawan yang  bertahan, namun masih cukup kesulitan dalam mendeteksi karyawan yang benar-benar akan keluar (dari 36 hanya 4 yang benar). Hal ini terjadi pada data HR karena alasan seseoran guntuk *resign*  sangat kompleks dan terkadang tidak sepenuhnya tercermin dalam data administratif. Model ini sudah bisa digunakan sebagai alat bantu *screening* leh tim HR

```python
# Menyiapkan data yang akan diprediksi (buang feature=NaN)
X_unseen = df_predict.drop(columns=["Attrition"])

# Encoding
X_unseen_encoded = pd.get_dummies(X_unseen, drop_first=True)

# Menyamakan jumlah dan urutan data dengan data train
# Jika ada NaT diisi dengan 0 supaya struktur tabel sama persis
X_unseen_encoded = X_unseen_encoded.reindex(columns=X_train.columns, fill_value=0)

# Prediksi
prediksi_baru = rf_model.predict(X_unseen_encoded)

# Memasukkan hasil tebakan ke dalam tabel aslinya
df_predict["Attrition"] = prediksi_baru

print("Tahap 6 Selesai! Mesin berhasil menebak nasib 412 karyawan.")
display(df_predict[['Age', 'Department', 'MonthlyIncome', 'Attrition']].head())
```
